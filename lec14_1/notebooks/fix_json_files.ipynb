```python
import json
import os

# Правильні дані для кожного файла
files_data = {
    "../data/few_files/1.json": [
        {
            "id": 1,
            "name": "John Doe",
            "details": {
                "age": 30,
                "address": {
                    "city": "New York",
                    "state": "NY"
                }
            }
        }
    ],
    "../data/few_files/2.json": [
        {
            "id": 2,
            "name": "Jane Smith",
            "details": {
                "age": 25,
                "address": {
                    "city": "Los Angeles",
                    "state": "CA"
                }
            }
        }
    ],
    "../data/few_files/3.json": [
        {
            "id": 3,
            "name": "Michael Johnson",
            "details": {
                "age": 40,
                "address": {
                    "city": "Chicago",
                    "state": "IL"
                }
            }
        }
    ]
}

# Створюємо файли
for file_path, records in files_data.items():
    # Створюємо директорію, якщо її нема
    os.makedirs(os.path.dirname(file_path), exist_ok=True)
    
    # Запис у форматі JSON Lines
    with open(file_path, "w", encoding="utf-8") as f:
        for record in records:
            json.dump(record, f)
            f.write("\n")
```
