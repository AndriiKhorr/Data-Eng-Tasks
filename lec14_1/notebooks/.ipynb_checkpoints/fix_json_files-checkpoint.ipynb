```python
import json
import glob

# Шлях до твоїх JSON файлів
files = glob.glob("../data/few_files/*.json")

for file_path in files:
    with open(file_path, "r", encoding="utf-8") as f:
        data = json.load(f)  # Завантажуємо масив обʼєктів з файлу

    # Переписуємо у формат JSON Lines (по одному обʼєкту на рядок)
    with open(file_path, "w", encoding="utf-8") as f:
        for record in data:
            json.dump(record, f)
            f.write("\n")
```
